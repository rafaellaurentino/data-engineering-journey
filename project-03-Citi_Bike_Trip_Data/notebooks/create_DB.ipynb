```python
import duckdb
import os

os.makedirs('../database', exist_ok=True)
con = duckdb.connect('../database/citibike.duckdb')
```

```python
con.execute("""
CREATE OR REPLACE TABLE trips_raw AS
SELECT *
FROM read_csv_auto(
    '../data/*.csv',
    union_by_name = true,
    filename = true,
    types = {'start_station_id': 'VARCHAR', 'end_station_id': 'VARCHAR'}
)
""")
```

```python
con.close()
```
